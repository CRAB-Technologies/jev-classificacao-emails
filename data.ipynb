```python
import pandas as pd

df = pd.read_csv("data/raw/exemplos_pseudonymized(Sheet1).csv", encoding="cp1252", sep=";")

print(df.shape)
df = df.drop_duplicates(subset=["Subject_masked", "BodyPlainText_masked"])
print(df.shape)  # deve dar (587, 6)
```

```python
from classify import clean_body

# Remove o aviso "Este e-mail vem de fora da organização. / Não clique em links..." de todos os bodies
df["BodyPlainText_masked"] = df["BodyPlainText_masked"].fillna("").map(clean_body)

avisos = df["BodyPlainText_masked"].str.contains("clique em links nem abra", case=False).sum()
print(f"Emails ainda com o aviso: {avisos}")
print(f"Emails que ficaram vazios (só tinham o aviso): {(df['BodyPlainText_masked'] == '').sum()}")
df["BodyPlainText_masked"].head()

df.to_csv("data/clean.csv", sep=";", index=False, encoding="utf-8")
```
